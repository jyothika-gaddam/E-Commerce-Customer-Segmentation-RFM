# E-Commerce Customer Segmentation (RFM Analysis)

This notebook runs the included demo dataset through the project pipeline. For the full UCI Online Retail dataset, update `input_path` below.

## 1. Imports and input path

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if not (ROOT / 'src').exists() and (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))
from rfm_analysis import build_rfm, save_outputs
input_path = ROOT / 'data' / 'sample_online_retail.csv'
output_dir = ROOT / 'outputs'
print('Input:', input_path)
```

## 2. Build RFM table

- **Recency:** days since last purchase; lower is better.
- **Frequency:** number of distinct invoices/orders.
- **Monetary:** total transaction revenue.

```python
rfm, reference_date = build_rfm(input_path)
print('Reference date:', reference_date.date())
print('Customers:', len(rfm))
rfm.head(10)
```

## 3. Segment summary

```python
summary = rfm.groupby('Segment').agg(
    Customer_Count=('CustomerID', 'nunique'),
    Total_Revenue=('Monetary', 'sum'),
    Average_Recency_Days=('Recency', 'mean'),
    Average_Orders=('Frequency', 'mean'),
    Average_Customer_Spend=('Monetary', 'mean')
).reset_index().sort_values('Total_Revenue', ascending=False)
summary
```

## 4. Save deliverables

The CSV tables and charts are saved in the `outputs/` directory.

```python
save_outputs(rfm, reference_date, output_dir)
```
